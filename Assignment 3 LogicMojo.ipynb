Q6. Handling Missing Data (Critical ML Step)

Context: Models cannot handle null values.

Dataset: Titanic

Tasks:

1. Fill missing age using median
2. Fill missing embarked using mode
3. Drop deck column
4. Explain:
Why different strategies are used for different columns

```python
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = pd.DataFrame(titanic)

print("Fill missing age using median")
df["age"] = df["age"].fillna(df["age"].median())

print("Fill missing embarked using mode")
df["embarked"] = df["embarked"].fillna(df["embarked"].mode())

print("Drop deck column")
df.drop(columns = ["deck"], inplace=True)

print("Different strategies are used for different columns because data types are different.\
Numerical columns can be imputed using mean or median, whereas categorical columns require mode")
```

Q8. Group-Based Insights (Customer Segmentation Logic)

Context: ML models often depend on aggregated patterns.

Dataset: Tips

Tasks:

1. Compute:
Average tip by day
Total bill by gender
2. Create new feature:
tip_percentage
3. Identify:
Which day has highest tipping behavior
4. Explain:
How this helps in segmentation models

```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("tips")

print("Average tip by day")
avg = df.groupby("day", observed=True)["tip"].mean()
print(avg)

print("Total bill by gender")
sum = df.groupby("sex", observed=True)["total_bill"].sum()
print(sum)

print("tip_percentage")
df["tip_percentage"] = (df["tip"]/df["total_bill"])*100

print("Which day has highest tipping behavior")
highest_tipping_day = df.groupby("day", observed=True)["tip_percentage"].mean()
print(highest_tipping_day)

print("Feature engineering and aggregation directly improve customer segmentation by transforming raw, transactional data into meaningful, behavioral, \
and high-value insights that machine learning models can easily interpret")
```

Q11. End-to-End ML Preprocessing Pipeline (Titanic)

Context: Prepare dataset for ML model training

Tasks:

1. Load dataset
2. Handle missing values
3. Encode categorical variables (sex, embarked)
4. Normalize numerical features
5. Convert to NumPy array

Output:

Final feature matrix ready for model training

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
import seaborn as sns

# Load dataset
df = sns.load_dataset("titanic")

# Handle missing values
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])

# Encode categorical variables (sex, embarked)
df = pd.get_dummies(df, columns=["sex", "embarked"], drop_first=True)

# Normalize numerical features
scaler = StandardScaler()

num_cols = ["age", "fare"]
df[num_cols] = scaler.fit_transform(df[num_cols])

# Convert to NumPy array
array = df.to_numpy()
print(array)
```

Q13. Custom Feature Transformation using apply()

Context: Real ML requires custom logic transformation.

Dataset: Titanic

Tasks:

1. Create age_group:
Child / Adult / Senior
2. Use .apply()
3. Compute survival rate per group
4. Interpret:
Which segment has highest survival likelihood

```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("titanic")
print(df)

# Create age_group: Child / Adult / Senior, Use .apply()
def categorize_age(age):
    if pd.isna(age):
        return "Unknown"
    elif age < 18:
        return "Child"
    elif age <= 60:
        return "Adult"
    else:
        return "Senior"

df['age_group'] = df['age'].apply(categorize_age)

# Compute survival rate per group
survival_rate = df.groupby('age_group')["survived"].mean()
print(f"Survival rate: {survival_rate}")

# Which segment has highest survival likelihood
print("Child have highest survival likelihood")
```

```python
# Q14 Error Analysis Simulation (Model Evaluation Thinking)
import numpy as np

# Generate 1000 random error values using NumPy
errors = np.random.randn(1000)
print(errors)

# Compute Mean and Standard Deviation
mean = np.mean(errors)
print(f"Mean: {mean}")
std = np.std(errors)
print(f"Standard Deviation: {std}")

# Identify outliers: Using mean ± 2*std
lowerbound = mean - 2*std
upperbound = mean + 2*std

outliers = errors[(errors < lowerbound) | (errors > upperbound)]
print(f"Outliers: {outliers}")

# What high variance means in model performance
print("High variance in machine learning means a model is too sensitive to small fluctuations in the training data, leading to overfitting. \
It indicates the model has learned noise rather than patterns, resulting in excellent performance on training data but poor performance on new, unseen test data")
###########################################################################################################################################################
# Q15 Business + ML Insight Problem (Tips Dataset)
import pandas as pd
import seaborn as sns

# tip percentage
df = sns.load_dataset("tips")
df["tip_percentage"] = ((df["tip"])/(df["total_bill"]))*100
print(df)
# day and time
df.groupby("day", observed=True)["tip_percentage"]
df.groupby("time", observed=True)["tip_percentage"]

# Highest revenue segment
df.loc[df["tip_percentage"].idxmax()]
```
